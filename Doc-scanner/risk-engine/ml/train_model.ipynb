```python

company_df.columns = company_df.columns.str.strip()

# Everything except 'class' should be numeric
# Force convert — anything that can't convert becomes NaN
for col in company_df.columns:
    if col != 'class':
        company_df[col] = pd.to_numeric(company_df[col], errors='coerce')

# Convert class column separately — strip spaces first
company_df['class'] = company_df['class'].astype(str).str.strip()
company_df['class'] = pd.to_numeric(company_df['class'], errors='coerce')

print("Data types after conversion:")
print(company_df.dtypes.value_counts())
print("\nMissing values per column:")
print(company_df.isnull().sum().sort_values(ascending=False).head(20))
print("\nClass distribution:")
print(company_df['class'].value_counts())
```

```python
features = [
    'net_profit_over_total_assets',
    'total_liabilities_over_total_assets',
    'working_capital_over_total_assets',
    'current_assets_over_short_term_liabilities',
    'retained_earnings_over_total_assets',
    'EBIT_over_total_assets',
    'book_value_of_equity_over_total_liabilities',
    'sales_over_total_assets',
    'equity_over_total_assets',
    'gross_profit_over_sales',
    'net_profit_over_sales',
    'profit_on_operating_activities_over_total_assets',
]

# Keep only rows that have at least half the features
company_df_clean = company_df[features + ['class']].dropna(thresh=len(features)//2)

print(f"Rows after cleaning: {len(company_df_clean)}")
print(f"Original rows: {len(company_df)}")
```

```python
def compute_tax_risk(row):
    score = 0

    val = row['total_liabilities_over_total_assets']
    if pd.notna(val):
        if float(val) > 0.7:   score += 25
        elif float(val) > 0.5: score += 10

    val = row['net_profit_over_sales']
    if pd.notna(val):
        if float(val) < 0:     score += 25
        elif float(val) < 0.05: score += 10

    val = row['profit_on_operating_activities_over_total_assets']
    if pd.notna(val):
        if float(val) < 0.02:  score += 20

    val = row['current_assets_over_short_term_liabilities']
    if pd.notna(val):
        if float(val) < 1.0:   score += 15

    val = row['class']
    if pd.notna(val) and float(val) == 1:
        score += 15

    return min(score, 100)

company_df_clean = company_df_clean.copy()
company_df_clean['tax_risk_score'] = company_df_clean.apply(compute_tax_risk, axis=1)

def to_label(score):
    if score <= 30:   return 'Low'
    elif score <= 60: return 'Medium'
    else:             return 'High'

company_df_clean['risk_label'] = company_df_clean['tax_risk_score'].apply(to_label)

print("Risk label distribution:")
print(company_df_clean['risk_label'].value_counts())
print("\nSample:")
print(company_df_clean[['tax_risk_score', 'risk_label']].head(10))
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
from sklearn.impute import SimpleImputer

X = company_df_clean[features]
y = company_df_clean['risk_label']

# Fill remaining missing values with column median
imputer = SimpleImputer(strategy='median')
X_imputed = imputer.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(
    X_imputed, y,
    test_size=0.2,
    random_state=42,
    stratify=y
)

print(f"Training: {len(X_train)} | Test: {len(X_test)}")

model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    class_weight='balanced'
)
model.fit(X_train, y_train)

y_pred = model.predict(X_test)
print("\n✅ Model Performance:")
print(classification_report(y_test, y_pred))
```

```python
import joblib, os

os.makedirs('../../models', exist_ok=True)

joblib.dump(model,    '../../models/tax_risk_model.pkl')
joblib.dump(imputer,  '../../models/tax_risk_imputer.pkl')
joblib.dump(features, '../../models/model_features.pkl')

print("✅ Saved to risk-engine/models/")
print("   tax_risk_model.pkl")
print("   tax_risk_imputer.pkl")
print("   model_features.pkl")
```

```python
import joblib
import numpy as np
import pandas as pd

# Load everything back
loaded_model    = joblib.load('../../models/tax_risk_model.pkl')
loaded_imputer  = joblib.load('../../models/tax_risk_imputer.pkl')
loaded_features = joblib.load('../../models/model_features.pkl')

# Simulate what your doc-scanner will send
# These are ratios computed from a real balance sheet + income statement
test_company = {
    'net_profit_over_total_assets':                       0.02,
    'total_liabilities_over_total_assets':                0.75,  # high debt
    'working_capital_over_total_assets':                 -0.05,  # negative working capital
    'current_assets_over_short_term_liabilities':         0.80,  # below 1 = bad
    'retained_earnings_over_total_assets':                0.01,
    'EBIT_over_total_assets':                             0.03,
    'book_value_of_equity_over_total_liabilities':        0.30,
    'sales_over_total_assets':                            0.50,
    'equity_over_total_assets':                           0.25,
    'gross_profit_over_sales':                            0.08,
    'net_profit_over_sales':                              0.01,
    'profit_on_operating_activities_over_total_assets':   0.01,
}

# Convert to dataframe
input_df = pd.DataFrame([test_company])

# Impute & predict
input_imputed   = loaded_imputer.transform(input_df)
prediction      = loaded_model.predict(input_imputed)[0]
probabilities   = loaded_model.predict_proba(input_imputed)[0]
classes         = loaded_model.classes_

print("✅ Model loaded and working!")
print(f"\nTest company risk level: {prediction}")
print("\nConfidence scores:")
for cls, prob in zip(classes, probabilities):
    bar = '█' * int(prob * 30)
    print(f"  {cls:8s}: {prob*100:5.1f}%  {bar}")
```
